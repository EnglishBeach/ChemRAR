# Конвертация весов модели SCScore

Взята модель full_reaxys_model_1024bool из https://github.com/connorcoley/scscore/tree/master


````python
import gzip
import json
import pickle


def convert_scscore_weights(json_gz_path: str, pickle_path: str) -> None:
    with gzip.open(json_gz_path, "rt", encoding="utf-8") as f:
        flat_vars = json.load(f)
    weights = flat_vars[0::2]
    biases = flat_vars[1::2]
    with open(pickle_path, "wb") as f:
        pickle.dump((weights, biases), f)
````

````python
convert_scscore_weights(
    "data/scscore/full_reaxys_model_1024bool/model.ckpt-10654.as_numpy.json.gz",
    "data/scscore/scscore_1024bool.pickle",
)
````
